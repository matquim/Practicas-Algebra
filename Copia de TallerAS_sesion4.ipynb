## 4.2 Producto punto

Esta operación de define como:

$\mathbf{u} \cdot \mathbf{v} = \displaystyle{\sum\limits_{k=1}^{n}} u_k v_k$

**Ejemplo**: $(2,-3)\cdot (5,3) = 1$



````python
u = [2, -3]
v = [5, 3]

dot = u[0]*v[0] + u[1]*v[1]
dot2 = sum(u[i] * v[i] for i in range(len(u)))

print(dot)
print(dot2)
````

También podemos hacerlo definiendo una función:

````python
def producto_punto(v, w):
    # Verificamos que los vectores tengan el mismo número de componentes
    if len(v) != len(w):
        print("Error: los vectores deben tener el mismo número de componentes")
        return None

    resultado = 0

    for i in range(len(v)):
        resultado = resultado + v[i] * w[i]
    return resultado

producto_punto(u,v)
````

## 4.4 Ángulo entre vectores

El coseno del angulo entre dos vectores se define como:

$$\cos{\theta}=\dfrac{\mathbf{u}\cdot\mathbf{v}}{||\mathbf{u}||\; ||\mathbf{v}||}$$

por lo que:

$$\theta=\text{arccos}\left(\dfrac{\mathbf{u}\cdot\mathbf{v}}{||\mathbf{u}||\; ||\mathbf{v}||}\right)$$

**Ejemplo** Calcule el ángulo entre los vectores $\mathbf{a}$ y $\mathbf{b}$ que definió en el ejercicio 4.3.2 para calcular el área del triángulo.

Suponiendo que $\mathbf{a}=\overline{P_1P_2}$ y que $\mathbf{b}=\overline{P_1P_3}$ , entonces:

````python
import math as mt
p1 = [0, 1, 3]
p2 = [-1, 2, 0]
p3 = [0, 4, 1]

a = fv.sumar_vectores(p2,fv.escalar_por_vector(-1,p1))
b = fv.sumar_vectores(p3,fv.escalar_por_vector(-1,p1))

theta = mt.acos(producto_punto(a,b)/(fv.norma(a)*fv.norma(b)))
print(theta) # resultado en radianes
print(mt.degrees(theta)) # resultado en grados
````

## Complemento 2: Operaciones vectoriales con NumPy

La gran mayoría de operaciones que realizamos en esta sesión y la anterior se pueden llevar a cabo usando la biblioteca NumPy.

````python
import numpy as np

u = np.array([2, -3, 1])
v = np.array([4, 5, -2])

print("Suma:", u + v)
print("Resta:", u - v)
print("Escalar por vector:", 3*u)
print("Producto punto:", np.dot(u, v))
print("Producto cruz:", np.cross(u, v))
print("Norma de u:", np.linalg.norm(u))
print("Distancia entre u y v:", np.linalg.norm(u - v))
print("Producto componente a componente:", u * v)
print("División componente a componente:", u / v)
````

Es importante notar que no existen funciones en NumPy para todas las operaciones que realizamos, como ejemplos están: el cálculo de el ángulo entre vectores o el cálculo del área de un paralelogramo. Sin embargo se pueden crear códigos simples a partir de las funciones de NumPy para calcularlas, por ejemplo, el cálculo del ángulo entre vectores con numpy se puede hacer mediante:

````python
theta = (np.arccos(np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))))
print(theta)
````
