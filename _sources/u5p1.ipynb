\end{equation}

Por ejemplo
\begin{equation}
    w = 2 \begin{pmatrix}5 \\ 3 \\ 2\end{pmatrix} - \begin{pmatrix}1 \\ 3 \\ -2\end{pmatrix} = \begin{pmatrix}9 \\ 3 \\ 6\end{pmatrix}
\end{equation}

`````python
import numpy as np
import matplotlib.pyplot as plt

v1 = np.array([2, 1])

a_values = np.linspace(-3, 3, 500)

x = []
y = []

for a in a_values:
    v = a*v1
    x.append(v[0])
    y.append(v[1])


plt.scatter(x, y, s=2, color="gray")
plt.quiver(0, 0, v1[0], v1[1],
           angles="xy",
           scale_units="xy",
           scale=1,
           color="green")

plt.text(v1[0] + 0.1 , v1[1] - 0.5 , r'(2,1)',
         fontsize=12, color='green')

plt.axis("equal")
plt.grid()
plt.show()
`````
