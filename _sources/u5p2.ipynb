# Base y dimensión

### Ejemplo de Base y dimensión en $\mathbb{R}^3$

1. Verificar si los vectores $\mathbf{u}$, $\mathbf{v}$ y $\mathbf{w}$ forman una base de $\mathbb{R}^3$

$$
\mathbf{u}=
\begin{pmatrix}
1\\
2\\
0
\end{pmatrix},
\qquad
\mathbf{v}=
\begin{pmatrix}
0\\
1\\
1
\end{pmatrix},
\qquad
\mathbf{w}=
\begin{pmatrix}
2\\
0\\
1
\end{pmatrix}
$$

Usando Python podemos hacer los siguiente

* Construir una matriz $A$ cuyas columnas sean los vectores $\mathbf{u}$, $\mathbf{v}$ y $\mathbf{w}$

* Usar Python para calcular el determinante de $A$ y determinar si los vectores efectivamente forman una base del espacio $\mathbb{R}^3$


$\mathbf{u}$, $\mathbf{v}$ y $\mathbf{w}$ son linealmente independientes y por lo tanto forman una base de $\mathbb{R}^3$


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
u =  np.array([1, 2, 0])
v =  np.array([0, 1, 1])
w =  np.array([2, 0, 1])

A = np.column_stack((u, v, w))

det_A = np.linalg.det(A)

print(det_A)
```


2. Sea $
\mathbf{p}=
\begin{pmatrix}
3\\
4\\
2
\end{pmatrix}
$

Exprese el vector $\mathbf{p}$ como una combinación lineal de los vectores $\mathbf{u}$, $\mathbf{v}$ y $\mathbf{w}$, es decir

$ \mathbf{p} = \alpha_1\mathbf{u} + \alpha_2\mathbf{v} + \alpha_3\mathbf{w}$


```python
d = np.array([3, 4, 2])

coeficientes =  np.linalg.solve(A, d)

#print(coeficientes)
print(f" p = {coeficientes[0]}u + {coeficientes[1]}v {coeficientes[2]}w")
```

### Espacios generados

Considere el vector

$$
\mathbf{v_1}=
\begin{pmatrix}
2\\
1
\end{pmatrix}
$$

El espacio generado por $\mathbf{v_1}$ está dado por

$$
\operatorname{gen}\left\{\mathbf{v_1}
\right\}
=
\left\{\mathbf{v}: \mathbf{v} = a\mathbf{v_1}, a\in\mathbb{R}
\right\}
$$


```python
v1 = np.array([2, 1])

a_values = np.linspace(-3, 3, 500)

x = []
y = []

for a in a_values:
    v = a*v1
    x.append(v[0])
    y.append(v[1])


plt.scatter(x, y, s=2, color="gray")
plt.quiver(0, 0, v1[0], v1[1],
           angles="xy",
           scale_units="xy",
           scale=1,
           color="green")

plt.text(v1[0] + 0.1 , v1[1] - 0.5 , r'(2,1)',
         fontsize=12, color='green')

plt.axis("equal")
plt.grid()
plt.show()
```

1. Considere los vectores

$$
\mathbf{w_1}=
\begin{pmatrix}
1\\
0
\end{pmatrix},
\qquad
\mathbf{w_2}=
\begin{pmatrix}
0\\
1
\end{pmatrix}
$$

El espacio generado por $\mathbf{w_1}$ y $\mathbf{w_2}$ está dado por

$$
\operatorname{gen}\left\{\mathbf{w_1},\mathbf{w_2}\right\}
=
\left\{
\mathbf{w}: \mathbf{w}=a\mathbf{w_1}+b\mathbf{w_2},
\;a,b\in\mathbb{R}
\right\},
$$

en donde

$$
a\mathbf{w_1}+b\mathbf{w_2}
=
a
\begin{pmatrix}
1\\
0
\end{pmatrix}
+
b
\begin{pmatrix}
0\\
1
\end{pmatrix}
=
\begin{pmatrix}
a\\
b
\end{pmatrix}
$$



```python
# Escriba su código aquí
```

```python
w1 = np.array([1, 0])
w2 = np.array([0, 1])

x = []
y = []

for a in np.linspace(-3, 3, 100):
    for b in np.linspace(-3, 3, 100):
        w = a*w1 + b*w2
        x.append(w[0])
        y.append(w[1])

plt.scatter(x, y, color="gray", s=1)

plt.quiver(0, 0, w1[0], w1[1],
           angles="xy",
           scale_units="xy",
           scale=1,
           color="red")

plt.quiver(0, 0, w2[0], w2[1],
           angles="xy",
           scale_units="xy",
           scale=1,
           color="blue")

plt.axis("equal")
plt.grid()
plt.show()
```

```python
# Escriba su código aquí
```

```python
w1 = np.array([1, 0, 0])
w2 = np.array([0, 1, 1])

x = []
y = []
z = []

for a in np.linspace(-3, 3, 50):
    for b in np.linspace(-3, 3, 50):
        w = a*w1 + b*w2
        x.append(w[0])
        y.append(w[1])
        z.append(w[2])

fig = plt.figure()

ax = fig.add_subplot(projection='3d')

# Puntos del plano
ax.scatter(x, y, z, color='gray', s=2)

# Vectores generadores
ax.quiver(0, 0, 0,
          w1[0], w1[1], w1[2],
          color='red')

ax.quiver(0, 0, 0,
          w2[0], w2[1], w2[2],
          color='blue')

L = 4

ax.quiver(0, 0, 0, L, 0, 0, arrow_length_ratio=0.05, color='black')
ax.quiver(0, 0, 0, 0, L, 0, arrow_length_ratio=0.05, color='black')
ax.quiver(0, 0, 0, 0, 0, L, arrow_length_ratio=0.05, color='black')


# Etiquetas
ax.text(w1[0] + 0.1, w1[1], w1[2],
        '(1,0,0)', color='red')

ax.text(w2[0], w2[1] + 0.1, w2[2],
        '(0,1,1)', color='blue')

# Ejes
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()
```
