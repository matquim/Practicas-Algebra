# Números complejos

Los números complejos en Python se pueden declarar usando la función complex() o directamente utilizando la notación $a + bj$.

Importante: La unidad imaginaria en Python se denota con $j$, i.e. $j^2=-1$

Ejemplo

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
z =  1 + 2j

w = complex(1, 2)

print(z)
print(w)
print(z == w)
```

### Conversión de flotantes/enteros a complejos

```python
var1 =  1
var2 = np.pi

complejo1 = complex(var1)
complejo2 = complex(var2)

print(complejo1, complejo2)
```

Para acceder a la parte real e imaginaria de un número complejo utilizando los atributos .real y .imag, respectivamente:

Ejemplo


```python
z_parte_real = z.real
z_parte_imag = z.imag

print(f"La parte real de z es: {z_parte_real} y la parte imaginaria es: {z_parte_imag}")
```

### Operaciones con complejos
### Suma

```python
z1 = 3 + 4j
z2 = 1 - 2j
print(z1)
print(z2)

z3 = z1 + z2

print(z3)
```

### Producto

```python
z4 =  z1*z2
print(z4)
```

### Complejo conjugado

```python
print(z1.conjugate())
print(z2.conjugate())
```

### División

```python
z5 = z1/z2
print(z5)
```

### Potencia

```python
z2**3
```

La función `np.angle()` devuelve el argumento principal del número complejo

$$
-\pi < \arg(z) \leq \pi,
$$

o bien

$$
-180^\circ < \arg(z) \leq 180^\circ
$$


```python
z1_angle_rad = np.angle(z1)
z2_angle_rad = np.angle(z2)

z1_angle_deg = np.degrees(z1_angle_rad)
z2_angle_deg = np.degrees(z2_angle_rad)

print("Argumento en radianes")
print(f"arg(z1) = {z1_angle_rad:8.4f}")
print(f"arg(z2) = {z2_angle_rad:8.4f}")

print("Argumento en grados")
print(f"arg(z1) = {z1_angle_deg:8.4f}")
print(f"arg(z2) = {z2_angle_deg:8.4f}")
```
